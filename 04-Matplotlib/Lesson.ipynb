```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([23, 24, 22, 25, 29])

plt.plot(x, y, marker='o', color='blue', markersize=8, label='Data Points')

plt.xlabel('Number', fontsize=20, color='green', loc='center', labelpad=15)
plt.ylabel('Year', fontsize=20, color='red', loc='center', labelpad=15)
plt.title('Simple Plot', fontsize=25, color='blue', loc='center', pad=20)
plt.grid(color='gray', linestyle='--', linewidth=0.5)

plt.show()
```

```python
# Customizing the plot with different styles and markers

plt.plot(x, y,  marker='o', color='blue', markersize=8, label='Data Points')
```

```python
# Bar Chart

x = np.array(['A', 'B', 'C', 'D', 'E'])
y = np.array([2, 4, 3, 6, 3])

plt.bar(x, y, color='blue')

plt.show()
```

```python
categories = ["Freshman", "Sophomore", "Junior", "Senior"]
values = np.array([300, 250, 275, 225])

plt.pie(values, labels=categories, autopct='%1.1f%%', startangle=90, colors=['#ff9999','#66b3ff','#99ff99','#ffcc99'])

plt.title('Student Distribution by Year', fontsize=20, color='blue', loc='center', pad=20)

plt.show()
```

```python
# Scatter Plot

x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y, color='blue')
plt.title('Scatter Plot', fontsize=20, color='red', loc='center', pad=20)

plt.show()
```

```python
# Histogram

x = np.random.normal(0, 1, 500)
y = np.random.normal(0, 1, 500)

plt.hist(x, bins=30, alpha=0.5, label='X', color='blue')
plt.hist(y, bins=30, alpha=0.5, label='Y', color='red')
plt.title('Histogram', fontsize=20, color='green', loc='center', pad=20)
plt.legend()

plt.show()
```

```python
# Subplots

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

x = np.array([1, 2, 3, 4, 5])
y1 = np.array([2, 4, 6, 8, 10])
y2 = np.array([1, 3, 5, 7, 9])

ax1.plot(x, y1, color='blue', label='y = 2x')
ax1.set_title('Line Plot', fontsize=15, color='blue', loc='center', pad=10)
ax1.set_xlabel('X-axis', fontsize=12, color='green', loc='center', labelpad=5)
ax1.set_ylabel('Y-axis', fontsize=12, color='red', loc='center', labelpad=5)
ax1.grid(color='gray', linestyle='--', linewidth=0.5)

ax2.bar(x, y2, color='orange', label='y = x + 1')
ax2.set_title('Bar Chart', fontsize=15, color='orange', loc='center', pad=10)
ax2.set_xlabel('X-axis', fontsize=12, color='green', loc='center', labelpad=5)
ax2.set_ylabel('Y-axis', fontsize=12, color='red', loc='center', labelpad=5)
ax2.grid(color='gray', linestyle='--', linewidth=0.5)

plt.tight_layout()
plt.show()
```
