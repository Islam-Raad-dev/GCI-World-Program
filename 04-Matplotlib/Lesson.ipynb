```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([23, 24, 22, 25, 29])

plt.plot(x, y)

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Simple Plot')

plt.show()
```

```python
# Customizing the plot with different styles and markers

plt.plot(x, y,  marker='o', color='blue', markersize=8, label='Data Points')
```
