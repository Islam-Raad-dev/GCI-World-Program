```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('data.csv')

type_counts = df['Type1'].value_counts(ascending=True)

plt.barh(type_counts.index, type_counts.values, color='blue')
plt.title('Number of Pokemon by Type', fontsize=20, color='red', loc='center', pad=20)
plt.xlabel('Count', fontsize=14, color='black')
plt.ylabel('Type', fontsize=14, color='black')

plt.show()
```
